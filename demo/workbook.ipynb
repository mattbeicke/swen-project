```python
# Instructions:
# Run `pip install requests`

from json import loads, dumps
import requests

def displayRequest(r: requests.models.Response):
    print(f"Status code: {r.status_code}")
    try:
        print("Reponse data:", loads(r.content))
    except Exception:
        print("Raw response data:", r.content)
```

```python
req = requests.get("http://127.0.0.1:8080/cupboard/0")
displayRequest(req)
```

```python
need_data = {
    "name": "1000 Lines of Code",
    "description": "Created for some code project",
    "id": 100
}
req = requests.post("http://127.0.0.1:8080/cupboard/", dumps(need_data))
displayRequest(req)
```

```python
need_data = {
    "name": "1000 Lines of Code",
    "description": "Created for some code project",
    "id": 100
}
req = requests.post("http://127.0.0.1:8080/cupboard", dumps(need_data), headers={"Content-Type":"application/json"})
displayRequest(req)
```

```python
some_needs = [
    {"name": "100 Lines of Code", "description": "For a side-project", "id": 201},
    {"name": "100000 Lines of Code", "description": "For Google.com", "id": 202},
    {"name": "10 Lines of Code", "description": "For a bash script", "id": 203},
    {"name": "5 Boxes of Cereal", "description": "For a food pantry", "id": 204},
    {"name": "100$ Donation", "description": "Thanks for your support!", "id": 205},
]

for need in some_needs:
    requests.post("http://127.0.0.1:8080/cupboard", dumps(need), headers={"Content-Type":"application/json"})

req = requests.get("http://127.0.0.1:8080/cupboard/", data="Code", headers={"Content-Type":"application/json"})
displayRequest(req)
print("Needs found:")
for need in loads(req.content):
    print(need)
```

```python
login_data = {
    "username": "user",
    "password": "temp"
}

req = requests.post("http://127.0.0.1:8080/accounts/login", json=login_data, headers={"Content-Type":"application/json"})
displayRequest(req)
key = req.text
```
